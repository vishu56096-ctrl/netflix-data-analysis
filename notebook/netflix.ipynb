```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/netflix_data.csv")
```

```python
#data looking how its look 
df.head()
df.info()
df.shape
df.columns
df.isnull().sum()
```

```python
# missing value handle
df.drop_duplicates(inplace=True)
df['director'] = df['director'].fillna("Unknown")
df['cast'] = df['cast'].fillna("Not Available")
df['country'] = df['country'].fillna("Unknown")
df.dropna(subset=['rating', 'duration', 'date_added'], inplace=True)
```

```python
#movies show and tv show on nextflix data(count data )
sns.countplot(x='type', data=df)
plt.title("Movies vs TV Shows")
plt.xlabel("content type")
plt.ylabel("count")
plt.savefig("image/movies_tvshows.png")
plt.show()
```

```python
#Top 10 Countries Producing Netflix Content
plt.figure(figsize=(8,4))
top_country = df['country'].value_counts().head(10)
top_country.plot(kind='bar')
plt.title("top 10 Countries Producing Netflix ")
plt.xlabel("Country")
plt.tight_layout()
plt.ylabel("Number of shows/movies")
plt.savefig("image/top10_country.png", dpi=100)
plt.show()
```

```python
#Netflix Content Release Trend
sns.histplot(df['release_year'], bins=20)
plt.title("Netflix Content Release Trend")
plt.xlabel("release year")
plt.ylabel("number of content")
plt.savefig("image/netflix_release_trend.png")
plt.show()
```

```python
#top 10 netflix genres
top_genres = df['listed_in'].value_counts().head(10)
top_genres.plot(kind='barh')
plt.title("Top 10 Netflix Genres")
plt.ylabel("genres")
plt.xlabel("count")
plt.savefig("image/top10_netflix_genres.png", bbox_inches='tight', dpi=100)
plt.show()
```

```python
1. Netflix contains more Movies than TV Shows.

2. United States dominates Netflix content production.

3. Most content was released after 2015.

4. Drama and Comedy are highly popular genres.
```
